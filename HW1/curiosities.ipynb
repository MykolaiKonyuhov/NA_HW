\begin{equation*}
f(x):=x^3−6 a x^2+12 a^2 x-8 a^3=(x−2a)^3,\qquad\qquad a = \frac{1}{2}\cdot 10^7 -1=4 999 999,\qquad f(10^7)=8.
\end{equation*}

```julia
a = 4999999

f(x) = x^3−6*a*x^2+12*a^2*x-8*a^3
g(x) = (x−2a)^3
x₀ = 10.0^7 

f(x₀), g(x₀)
```

## Example

\begin{equation*}
f(x) = \frac{1}{x-\sqrt{x^2-1}} = \frac{x+\sqrt{x^2-1}}{(x+\sqrt{x^2-1})(x-\sqrt{x^2-1})}=x+\sqrt{x^2-1}
\end{equation*}

```julia
f(x) = 1/(x-sqrt(x^2-1))
g(x) = x+sqrt(x^2-1)

x = 10^7

f(x)-g(x)      # should be zero
```

## Example

\begin{equation*}
f(x) = x^9-18x^8+144x^7-672x^6+2016x^5-4032x^4+5376x^3-4608x^2+2304x-512 = (x-2)^9
\end{equation*}

```julia
using Plots:plot,plot!

f(x) = x^9-18x^8+144x^7-672x^6+2016x^5-4032x^4+5376x^3-4608x^2+2304x-512
g(x) = (x-2)^9

plot(f,xlimits=[1.925;2.075],label = "f")
plot!(g,label = "g")
```

```julia
using LinearAlgebra:norm

A = [1.2969 0.8648; .2161 .1441]
b = [.8642;.1440]
x = [2;-2]    # exact solution

norm(A*x-b)
```

```julia
Δb = 1e-8*[-1;1]
b̃ = b+Δb
x̃ = [.9911;-.4870]   # exact solution

norm(A*x̃-b̃)
```
